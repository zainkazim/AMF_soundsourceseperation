### Step 1: Load and Preprocess Audio
##### Load a stereo audio file (e.g., a WAV file) and ensure it has two channels.
##### Normalize the audio and check for silence or anomalies in the input.

```python
import numpy as np
import librosa
import soundfile as sf

# Load stereo audio file
audio_path = "/home/ZA/Music/Media project/AMF_MP/new wav files/M1F1-Alaw-AFsp.wav"
y, sr = sf.read(audio_path)

# Ensure it’s stereo
if y.ndim != 2 or y.shape[1] != 2:
    raise ValueError("The audio file must be stereo (2 channels).")

# Normalize the audio
y = y / np.max(np.abs(y))

# Print audio information
print(f"Sample Rate: {sr}")
print(f"Audio Shape: {y.shape}")  # (num_samples, 2 channels)
```

### Step 2: Compute the STFT
##### Compute the STFT for both channels to get their time-frequency representations.
##### Visualize the magnitude spectrograms of both channels for comparison.
######  Key Concepts:

###### The magnitude reflects the energy in each frequency bin.
###### The phase reflects the timing or shift of the frequency components.

```python
import matplotlib.pyplot as plt

# Compute STFT for both channels
n_fft = 1024   
hop_length = 512   
win_length = 1024   

stft_left = librosa.stft(y[:, 0], n_fft=n_fft, hop_length=hop_length, win_length=win_length)
stft_right = librosa.stft(y[:, 1], n_fft=n_fft, hop_length=hop_length, win_length=win_length)

print(f"\nwhat stft did to left channel", stft_left)
print(f"\nwhat stft did to right channel", stft_right)

# Compute magnitude and phase
magnitude_left, phase_left = np.abs(stft_left), np.angle(stft_left)
magnitude_right, phase_right = np.abs(stft_right), np.angle(stft_right)

print(f"\nphase_left values", phase_left)
print(f"\nphase_right values", phase_right)

# Visualize magnitude spectrograms
plt.figure(figsize=(12, 6))
plt.subplot(2, 1, 1)
plt.title("Left Channel - Magnitude Spectrogram")
plt.imshow(20 * np.log10(magnitude_left + 1e-10), aspect='auto', origin='lower', cmap='viridis')
plt.colorbar(format="%+2.0f dB")
plt.subplot(2, 1, 2)
plt.title("Right Channel - Magnitude Spectrogram")
plt.imshow(20 * np.log10(magnitude_right + 1e-10), aspect='auto', origin='lower', cmap='viridis')
plt.colorbar(format="%+2.0f dB")
plt.tight_layout()
plt.show()
```

##### Step 3: Analyze Phase Differences
##### Calculate the phase difference between the left and right channels.
##### This phase difference helps determine azimuth or directionality.
##### Key Idea:

###### For a sound source coming from a specific direction, the phase difference at each frequency bin will follow a predictable pattern.

```python
# Flatten the phase_diff array to 1D
phase_diff_flat = phase_diff.flatten()

# Plot histogram
plt.figure(figsize=(8, 5))
plt.title("Histogram of Phase Differences (Left - Right Channels)")
plt.hist(phase_diff_flat, bins=50, color='blue', alpha=0.7, edgecolor='black')
plt.xlabel("Phase Difference (radians)")
plt.ylabel("Frequency")
plt.grid(True)
plt.tight_layout()
plt.show()

plt.figure(figsize=(12, 6))

# Left channel phase
plt.subplot(2, 1, 1)
plt.title("Left Channel - Phase Spectrogram")
plt.imshow(phase_left, aspect='auto', origin='lower', cmap='twilight')
plt.colorbar(label="Phase (radians)")
plt.xlabel("Time Frames")
plt.ylabel("Frequency Bins")

# Right channel phase
plt.subplot(2, 1, 2)
plt.title("Right Channel - Phase Spectrogram")
plt.imshow(phase_right, aspect='auto', origin='lower', cmap='twilight')
plt.colorbar(label="Phase (radians)")
plt.xlabel("Time Frames")
plt.ylabel("Frequency Bins")

plt.tight_layout()
plt.show()

print(f"Phase Left (bin=100, time=10): {phase_left[100, 10]}")
print(f"Phase Right (bin=100, time=10): {phase_right[100, 10]}")
print(f"Phase Difference: {phase_diff[100, 10]}")


```

### Step 4: Identify and Filter Frequency Bins
###### Using the phase difference, identify which frequency bins belong to which source.
###### You can cluster the bins based on their phase difference patterns (e.g., using k-means or thresholding).

##### Key Steps:

###### Group bins with similar phase differences (e.g., bins with close azimuths).
###### Generate a binary mask for each group.
###### Apply the mask to the STFTs to separate sources.

```python
from sklearn.cluster import KMeans

# Flatten the phase difference matrix for clustering
flat_phase_diff = phase_diff.flatten()
flat_magnitude = magnitude_left.flatten()

# Stack features for clustering
features = np.stack((flat_phase_diff, flat_magnitude), axis=-1)

# Apply k-means clustering
num_sources = 2  # Assuming 2 sources
kmeans = KMeans(n_clusters=num_sources, random_state=0).fit(features)
labels = kmeans.labels_.reshape(phase_diff.shape)

# Visualize clusters
plt.figure(figsize=(10, 6))
plt.title("Clustered Frequency Bins (Sources)")
plt.imshow(labels, aspect='auto', origin='lower', cmap='tab10')
plt.colorbar(label="Cluster")
plt.xlabel("Time Frames")
plt.ylabel("Frequency Bins")
plt.tight_layout()
plt.show()



```

##### Step 5: Reconstruct Audio for Each Source
###### Use the masks generated from clustering to filter each source.
###### Reconstruct the audio for each source by applying the inverse STFT.


```python
import os  # For directory handling
import numpy as np
import librosa
import soundfile as sf

# Specify the output folder for saving reconstructed audio files
output_folder = "/home/ZA/Music/Media project/Reconstructed_audio"
os.makedirs(output_folder, exist_ok=True)  # Ensure the folder exists

# Initialize combined left and right channels
combined_left = None
combined_right = None

# Create masks for each source
masks = [(labels == i).astype(float) for i in range(num_sources)]

# Apply masks and combine reconstructed audio for all sources
for mask in masks:
    # Reconstruct the left channel
    masked_stft_left = stft_left * mask
    reconstructed_left = librosa.istft(masked_stft_left, hop_length=hop_length, win_length=win_length)

    # Reconstruct the right channel
    masked_stft_right = stft_right * mask
    reconstructed_right = librosa.istft(masked_stft_right, hop_length=hop_length, win_length=win_length)

    # Combine contributions from all sources
    if combined_left is None:
        combined_left = reconstructed_left
        combined_right = reconstructed_right
    else:
        combined_left += reconstructed_left
        combined_right += reconstructed_right

# Save the combined left channel (left audio only)
left_output_path = os.path.join(output_folder, "final_left_channel.wav")
sf.write(left_output_path, np.stack((combined_left, np.zeros_like(combined_left)), axis=-1), sr)
print(f"Saved left channel: {left_output_path}")

# Save the combined right channel (right audio only)
right_output_path = os.path.join(output_folder, "final_right_channel.wav")
sf.write(right_output_path, np.stack((np.zeros_like(combined_right), combined_right), axis=-1), sr)
print(f"Saved right channel: {right_output_path}")

# Save the stereo file with combined left and right channels
stereo_audio = np.stack((combined_left, combined_right), axis=-1)  # Combine into stereo
stereo_output_path = os.path.join(output_folder, "final_stereo.wav")
sf.write(stereo_output_path, stereo_audio, sr)
print(f"Saved stereo file: {stereo_output_path}")
```

```python
#Single wav files/ channel - 2 channels, eacvh to have left and right  both  
import os  # For directory handling
import numpy as np
import librosa
import soundfile as sf

output_folder = "/home/ZA/Music/Media project/Reconstructed_audio"
os.makedirs(output_folder, exist_ok=True)  # Ensure the folder exists

masks = [(labels == i).astype(float) for i in range(num_sources)]

reconstructed_audio_left = []
reconstructed_audio_right = []
for mask in masks:
    masked_stft_left = stft_left * mask
    reconstructed_left = librosa.istft(masked_stft_left, hop_length=hop_length, win_length=win_length)
    reconstructed_audio_left.append(reconstructed_left)

    masked_stft_right = stft_right * mask
    reconstructed_right = librosa.istft(masked_stft_right, hop_length=hop_length, win_length=win_length)
    reconstructed_audio_right.append(reconstructed_right)

# Save reconstructed audio as stereo files
for i, (left_audio, right_audio) in enumerate(zip(reconstructed_audio_left, reconstructed_audio_right)):
    stereo_audio = np.stack((left_audio, right_audio), axis=-1)  

    # Save the stereo audio
    output_path_stereo = os.path.join(output_folder, f"stereo_reconstructed_source_{i+1}.wav")
    sf.write(output_path_stereo, stereo_audio, sr)
    print(f"Saved stereo file: {output_path_stereo}")

    # Save isolated left channel
    left_output_path = os.path.join(output_folder, f"left_channel_source_{i+1}.wav")
    sf.write(left_output_path, np.stack((left_audio, np.zeros_like(left_audio)), axis=-1), sr)
    print(f"Saved left channel: {left_output_path}")

    # Save isolated right channel
    right_output_path = os.path.join(output_folder, f"right_channel_source_{i+1}.wav")
    sf.write(right_output_path, np.stack((np.zeros_like(right_audio), right_audio), axis=-1), sr)
    print(f"Saved right channel: {right_output_path}")


```
